```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = r"C:\Degree\MCA\Sem 3\Computer Vision\Noisy image.png"
img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if img is None:
    raise FileNotFoundError(f"Image not found: {image_path}")

def show_result(title, result):
    result = cv2.normalize(result, None, 0, 255, cv2.NORM_MINMAX)
    plt.imshow(result, cmap="gray")
    plt.title(title)
    plt.axis("off")
    plt.show()
```

```python
def smoothen_filter(Rimg, filter_kernel):
    """Apply an odd-sized 2D kernel to a grayscale image."""
    image = np.asarray(Rimg, dtype=np.float32)
    kernel = np.asarray(filter_kernel, dtype=np.float32)

    if image.ndim != 2:
        raise ValueError("Image must be grayscale.")
    if kernel.ndim != 2 or any(size % 2 == 0 for size in kernel.shape):
        raise ValueError("Kernel must be 2D with odd dimensions.")

    return cv2.filter2D(
        image,
        ddepth=cv2.CV_32F,
        kernel=kernel,
        borderType=cv2.BORDER_REFLECT
    )
```

```python
# 1. Mean filter
mean_kernel = np.ones((3, 3), dtype=np.float32) / 9
mean_result = smoothen_filter(img, mean_kernel)
show_result("Mean Filter", mean_result)

# 2. Median filter
median_result = cv2.medianBlur(img, 3)
show_result("Median Filter", median_result)

# 3. Min filter
morph_kernel = np.ones((3, 3), dtype=np.uint8)
min_result = cv2.erode(img, morph_kernel)
show_result("Min Filter", min_result)

# 4. Max filter
max_result = cv2.dilate(img, morph_kernel)
show_result("Max Filter", max_result)

# 5. Canny filter
canny_result = cv2.Canny(img, 100, 200)
show_result("Canny Filter", canny_result)

# 6. LoG filter: Gaussian smoothing followed by a Laplacian
gaussian_1d = cv2.getGaussianKernel(5, 1)
gaussian_kernel = gaussian_1d @ gaussian_1d.T
blurred = smoothen_filter(img, gaussian_kernel)

laplacian_kernel = np.array([
    [0, 1, 0],
    [1, -4, 1],
    [0, 1, 0]
], dtype=np.float32)

log_result = smoothen_filter(blurred, laplacian_kernel)
show_result("LoG Filter", np.abs(log_result))

# 7. DoG filter
small_1d = cv2.getGaussianKernel(7, 1)
large_1d = cv2.getGaussianKernel(7, 2)
dog_kernel = (small_1d @ small_1d.T) - (large_1d @ large_1d.T)
dog_result = smoothen_filter(img, dog_kernel)
show_result("DoG Filter", np.abs(dog_result))

# 8. Gaussian filter
gaussian_result = smoothen_filter(img, gaussian_kernel)
show_result("Gaussian Filter", gaussian_result)

# 9. Sobel filter
sobel_x_kernel = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
], dtype=np.float32)

sobel_y_kernel = np.array([
    [-1, -2, -1],
    [ 0,  0,  0],
    [ 1,  2,  1]
], dtype=np.float32)

sobel_x = smoothen_filter(img, sobel_x_kernel)
sobel_y = smoothen_filter(img, sobel_y_kernel)
sobel_result = np.sqrt(sobel_x**2 + sobel_y**2)
show_result("Sobel Filter", sobel_result)
```
