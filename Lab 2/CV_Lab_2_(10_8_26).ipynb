```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

#Read image
image=cv2.imread(r"/content/rgb_shapes_edges_202944_3840x2160.jpg")

#Convert BGR to RGB,Gray,Binary
rgb_image=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
gray_image=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
_,binary_image=cv2.threshold(gray_image,60,255,cv2.THRESH_BINARY)

plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.imshow(rgb_image)
plt.title("Color Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(gray_image,cmap="gray")
plt.title("Gray Image")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(binary_image,cmap="gray")
plt.title("Binary Image")
plt.axis("off")
```

```python
#Split RGB Channels
red=rgb_image[:,:,0]
green=rgb_image[:,:,1]
blue=rgb_image[:,:,2]

#Display the channels
plt.figure(figsize=(12,4))

plt.subplot(1, 4, 1)
plt.imshow(rgb_image)
plt.title("Original RGB")
plt.axis("off")

plt.subplot(1,4,2)
plt.imshow(red,cmap="gray")
plt.title("Red Channel")
plt.axis("off")

plt.subplot(1,4,3)
plt.imshow(green,cmap="gray")
plt.title("Green Channel")
plt.axis("off")

plt.subplot(1,4,4)
plt.imshow(blue,cmap="gray")
plt.title("Blue Channel")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Creat empty images
red_image=np.zeros_like(rgb_image)
green_image=np.zeros_like(rgb_image)
blue_image=np.zeros_like(rgb_image)

#Copy each channel
red_image[:,:,0]=rgb_image[:,:,0]
green_image[:,:,1]=rgb_image[:,:,1]
blue_image[:,:,2]=rgb_image[:,:,2]

#Display the images
plt.figure(figsize=(12,4))

plt.subplot(1, 4, 1)
plt.imshow(rgb_image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,4,2)
plt.imshow(red_image,cmap="gray")
plt.title("Red Component")
plt.axis("off")

plt.subplot(1,4,3)
plt.imshow(green_image,cmap="gray")
plt.title("Green Component")
plt.axis("off")

plt.subplot(1,4,4)
plt.imshow(blue_image,cmap="gray")
plt.title("Blue Component")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Create copies
without_red=rgb_image.copy()
without_blue=rgb_image.copy()
without_green=rgb_image.copy()

without_red[:,:,0]=0
without_blue[:,:,2]=0
without_green[:,:,1]=0

#Display the images
plt.figure(figsize=(10,5))

plt.subplot(1, 4, 1)
plt.imshow(rgb_image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,4,2)
plt.imshow(without_red)
plt.title("Without Red")
plt.axis("off")

plt.subplot(1,4,3)
plt.imshow(without_blue)
plt.title("Without Blue")
plt.axis("off")

plt.subplot(1,4,4)
plt.imshow(without_green)
plt.title("Without Green")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Read two images
img1 = cv2.imread("/content/download 1.png")
img2 = cv2.imread("/content/rgb_shapes_edges_202944_3840x2160.jpg")

# Ensure both images have identical dimensions
img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

# Blending parameters
alpha = 0.5
beta = 1 - alpha
gamma = 0

rows, cols, channels = img1.shape
blended = np.zeros((rows, cols, channels), dtype=np.uint8)

# Image addition / blending
for i in range(rows):
    for j in range(cols):
        for k in range(channels):
            pixel = alpha * img1[i, j, k] + beta * img2[i, j, k] + gamma
            blended[i, j, k] = np.uint8(np.clip(pixel, 0, 255))

# Convert BGR images to RGB for Matplotlib display
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2_rgb = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
blended_rgb = cv2.cvtColor(blended, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img1_rgb)
plt.title("Image 1")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(img2_rgb)
plt.title("Image 2")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(blended_rgb)
plt.title("Blended Image")
plt.axis("off")

plt.show()
```
